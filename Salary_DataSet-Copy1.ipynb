# Histogram : 

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
plt.hist(df['Salary'],bins=15)
plt.title('Yearly Salary Range')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.grid()
plt.show()
```

# Seaborn Histogram :

```python
import seaborn as sns

plt.figure(figsize=(8,6))
sns.histplot(df['Salary'],kde=True)
plt.title('Yearly Salary Range')
plt.show()
```

# Bar chart :

```python
Gender_count = df['Gender'].value_counts()
Gender_count.plot(kind='bar')
plt.title('Gender Count Chart')
plt.xlabel('Gender Type')
plt.ylabel('Count of Gender')
plt.grid()
plt.show()
```

# Seaborn Bar Chart :

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Gender',data=df)
plt.title('Gender Count Chart')
plt.show()
```

# Line Chart :

```python
Yearly_Salary = df.groupby('Years of Experience')['Salary'].mean()
plt.figure(figsize=(10,6))
plt.plot(Yearly_Salary.index,Yearly_Salary.values)
plt.title('Average Yearly Salary Increase')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

# Scatter Plot :

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Salary'],df['Age'])
plt.title('Age vs Salary')
plt.xlabel('Salary')
plt.ylabel('Age')
plt.show()
```

# Pie Chart :

### Gender Count Distribution Chart :  

```python
Gender_Count = df['Gender'].value_counts()
plt.figure(figsize=(7,7))
plt.pie(Gender_Count,labels=Gender_Count.index,autopct= '%1.1f%%')
plt.title('Gender Count Distribution')
plt.show()
```

### Top 10 Job Title Distribution Chart : 

```python
Count=df['Job Title'].value_counts().head(10).plot(kind = 'pie',autopct = '%1.1f%%',figsize=(7,7))
plt.title('Top 10 Job Titles Distribution')
plt.ylabel("")
plt.show()
```

### Education Level Distribution Chart : 

```python
Count=df['Education Level'].value_counts().plot(kind = 'pie',autopct = '%1.1f%%',figsize=(7,7))
plt.title('Education Level Distribution')
plt.ylabel("")
plt.show()
```

# HeatMap :

```python
plt.figure(figsize=(8,5))
sns.heatmap(df.select_dtypes(include='number').corr(),annot=True)
plt.show()
```
